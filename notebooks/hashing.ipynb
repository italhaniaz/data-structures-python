# <b> Hash Tables </b>

```python
class HashMap:
    def __init__(self):
        self.size = 10
        self.map = [None] * self.size
```

```python
def _get_hash(self, key):
    return key % self.size

HashMap._get_hash = _get_hash
```

```python
def add(self, key, value):
    key_hash = self._get_hash(key)
    key_value = [key, value]

    # insert or update = "upsert"

    self.map[key_hash] = [key_value]

    return True

HashMap.add = add
```

```python
def get(self, key):
    key_hash = self._get_hash(key)

    if self.map[key_hash] is not None:
        for pair in self.map[key_hash]:
            if pair[0] == key:
                return pair[1]

    raise KeyError(str(key))

HashMap.get = get
```

```python
def __str__(self):
    ret = ""

    for i, item in enumerate(self.map):
        if item is not None:
            ret += str(i) + ":" + str(item) + '\n'

    return ret

HashMap.__str__ = __str__
```

```python
h = HashMap()

h.add(17, "seventeen")
h.add(26, "twenty six")
h.add(35, "thirty five")

print(h)
```

## <b> Collisions and Avoidance </b>

```python
h = HashMap()

h.add(17, "seventeen")
h.add(26, "twenty six")
h.add(35, "thirty five")
h.add(25, "twenty five")
h.add(26, "twenty six updated")
h.add(887, "large number")

print(h)
```

```python
# we can see that there are the collisions in it 

# so we called it (hash collisions)
# - the differ nums produce the same output when pass through hash

# Solution one:-
# we can put the next updated value in the next box 

#     The problem with that is :- 
#         - if we need any retrival we have a large number of comparisions
#         - the values which are unique and in there box they also effect with that


# Solution two:-
# we can create a list instead of that in that box and append that one new number that list
```

```python
# so now the add function is 
def add(self, key, value):
    key_hash = self._get_hash(key)
    key_value = [key, value]

    if self.map[key_hash] is None:
        self.map[key_hash] = [key_value]
        return True

    else:
        # check if it's an update
        for pair in self.map[key_hash]:
            if pair[0] == key:
                print("updating: ", key)
                pair[1] = value
                return True

    # nope, it's collision insert it
    self.map[key_hash].append(key_value)

HashMap.add = add
```

```python
def delete(self, key):
    key_hash = self._get_hash(key)

    if self.map[key_hash] is None:
        raise KeyError(str(key))

    else:
        # One way
        # for i, item in enumerate(self.map[key_hash]):
        #     if item[0] == key:
        #         self.map[key_hash].pop(i)
        #         return True

        # second way by sir 

        for i in range(0, len(self.map[key_hash])):
            if self.map[key_hash][i][0] == key:
                self.map[key_hash].pop(i)
                return True

HashMap.delete = delete
```

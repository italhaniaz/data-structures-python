# <b> Problem Solving </b>

```python
class Node:
    def __init__(self, data=None):
        self.val = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
```

```python
def push(self, value):
    new_node = Node(value)

    if self.head is None:
        self.head = new_node
        return 

    last = self.head
    
    while last.next is not None:
        last = last.next
        
    last.next = new_node

LinkedList.push = push
```

```python
def remove(self, val):
    temp = self.head

    if temp is not None:
        if temp.val == val:
            self.head = temp.next
            return 

    while temp is not None:
        if temp.val == val:
            break

        prev = temp
        temp = temp.next

    if temp is None:
        return 

    prev.next = temp.next

LinkedList.remove = remove
```

```python

def __str__(self):
    ret_str = '['
    temp = self.head

    while temp is not None:
        ret_str += str(temp.val) + ', '
        temp = temp.next

    ret_str = ret_str.rstrip(', ')
    ret_str += ']'

    return ret_str

LinkedList.__str__ = __str__
```

```python
l = LinkedList()

l.push(1)
l.push(2)
l.push(3)
l.push(4)
l.push(5)

print(l)
```

## <b>Prob 1:</b> Getting minimum number

```python
def get_min(self):
    if self.head is None:
        raise Exception("Not have any value")

    l_min = self.head.val
    
    temp = self.head.next
    while temp is not None:
        if temp.val < l_min:
            l_min = temp.val

        temp = temp.next

    return l_min

LinkedList.get_min = get_min
```

```python
def find_min(self):
    if self.head is None: return None

    l_min = self.head.val
    l_min_i = 0

    temp = self.head.next
    counter = 1   # keep track of counter
    
    while temp is not None:
        if temp.val < l_min:
            l_min = temp.val
            l_min_i = counter

        counter += 1
        temp = temp.next

    return (l_min, l_min_i)

LinkedList.find_min = find_min
```

## <b>Prob 2:</b> Getting maximum number

```python
def get_max(self):
    if self.head is None:
        raise Exception("Not have any value")

    l_max = self.head.val
    
    temp = self.head.next
    while temp is not None:
        if temp.val > l_max:
            l_max = temp.val
        
        temp = temp.next

    return l_max

LinkedList.get_max = get_max
```
